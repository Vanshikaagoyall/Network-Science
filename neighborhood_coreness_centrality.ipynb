**DOCUMENTATION**

  
@brief: Calculates the normalized Neighbourhood Coreness Centrality (NCC) of a node.

@details: Calculates the k-shell (coreness) value of every node in the graph.Then, computes the NCC of the given node by summing the coreness
values of all its neighbours. The result is normalized by dividing the node's NCC by the maximum NCC value in the graph and rounded
to 4 decimal places.

@param G: A Graph on which NCC is calculated.

@param node: The node whose normalized NCC value is to be calculated.

@return: The normalized NCC value of the given node, rounded to
4 decimal places.

@note: The graph must be a simple graph without self-loops for nx.core_number(G) to work correctly. The graph must contain
at least one node with a positive NCC value to avoid division by zero.


**FUNCTION TO COMPUTE NORMALISED NEIGHBORHOOD CENTRALITY**

```python
import networkx as nx
import matplotlib.pyplot as plt


def neighborhood_coreness_centrality(G, node):

    # Calculate coreness of all nodes
    coreness = nx.core_number(G)

    # Calculate NCC of the given node
    ncc = 0

    for neighbor in G.neighbors(node):
        ncc = ncc + coreness[neighbor]

    # Calculate NCC of all nodes
    all_ncc = {}

    for n in G.nodes():

        total = 0

        for neighbor in G.neighbors(n):
            total = total + coreness[neighbor]

        all_ncc[n] = total

    # Find maximum NCC
    max_ncc = max(all_ncc.values())

    # Normalize NCC
    normalized_ncc = round(ncc / max_ncc,4)

    return normalized_ncc
```

**KARATE CLUB NETWORK**

```python
G = nx.karate_club_graph()

plt.figure(figsize=(8, 6))
position = nx.spring_layout(G, seed=42)
nx.draw(
    G,
    position,
    with_labels=True,
    node_size=500)

plt.title("Karate Club Network")
plt.show()
```

```python
# KARATE NETWORK

G = nx.karate_club_graph()

print("NCC VALUES - KARATE NETWORK")
print("----------------------------")

for node in G.nodes():
    value = neighborhood_coreness_centrality(G, node)
    print("Node:", node, "NCC:", value)
```

```python
# -----------------------------------------
# KARATE NETWORK
# -----------------------------------------

G = nx.karate_club_graph()


# Calculate NCC values
ncc_values = {}

for node in G.nodes():
    ncc_values[node] = neighborhood_coreness_centrality(G, node)


# Sort nodes by NCC
ranking = sorted(
    ncc_values,
    key=ncc_values.get,
    reverse=True
)


# Get Top 10
top_nodes = ranking[:10]

top_ncc = []

for node in top_nodes:
    top_ncc.append(ncc_values[node])


# -----------------------------------------
# PRINT TOP 10
# -----------------------------------------

print("TOP 10 NODES BY NCC")
print("-------------------")

for i in range(10):

    print(
        "Rank:", i + 1,
        "Node:", top_nodes[i],
        "NCC:", top_ncc[i]
    )


# -----------------------------------------
# BAR GRAPH
# -----------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    top_nodes,
    top_ncc
)

plt.xlabel("Node")
plt.ylabel("NCC Score")
plt.title("Top 10 Nodes by Neighbourhood Coreness Centrality")

plt.xticks(top_nodes)

plt.show()
```

**COMPARISION - KARATE VS DOLPHIN**

```python
# -----------------------------------------
# FUNCTION TO GET TOP 10 NCC NODES
# -----------------------------------------

def top_10_ncc(G):

    ncc_values = {}

    for node in G.nodes():
        ncc_values[node] = neighborhood_coreness_centrality(G, node)

    # Sort from highest NCC to lowest NCC
    ranking = sorted(
        ncc_values,
        key=ncc_values.get,
        reverse=True
    )

    top_nodes = ranking[:10]

    return top_nodes, ncc_values


# -----------------------------------------
# KARATE NETWORK
# -----------------------------------------

karate = nx.karate_club_graph()

karate_nodes, karate_ncc = top_10_ncc(karate)


# -----------------------------------------
# DOLPHIN NETWORK
# -----------------------------------------

dolphins = nx.read_gml("dolphins.gml")

dolphin_nodes, dolphin_ncc = top_10_ncc(dolphins)


# -----------------------------------------
# COMPARISON TABLE
# -----------------------------------------

print("\nTOP 10 Normalised NCC COMPARISON")
print("=" * 65)

print(
    "Rank".ljust(8),
    "Karate Node".ljust(15),
    "Karate NCC".ljust(15),
    "Dolphin Node".ljust(15),
    "Dolphin NCC"
)

print("-" * 65)

for i in range(10):

    karate_node = karate_nodes[i]
    dolphin_node = dolphin_nodes[i]

    print(
        str(i + 1).ljust(8),
        str(karate_node).ljust(15),
        str(karate_ncc[karate_node]).ljust(15),
        str(dolphin_node).ljust(15),
        dolphin_ncc[dolphin_node]
    )
```
